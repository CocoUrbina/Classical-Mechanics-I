# **Problema 3** Máquina de Atwood oscilante.

Encontremos el Lagrangiano:

$$  \mathcal{L} = T_1 + T_2 - V_1 - V_2$$

$T_1 = \frac{1}{2} M \dot{r}^2$

$T_2 = \frac{1}{2} m \dot{r}^2 + \frac{1}{2}m v_T ^2, \quad v_T = r\omega = r \dot{\theta}$

$T_2 = \frac{1}{2} m \dot{r}^2 + \frac{1}{2}m r^2 \dot{\theta}^2$

$V_1 = Mgr$

$V_2 = mgr\cos\theta$

\begin{align*}
\mathcal{L} &= T_1 + T_2 - V_1 - V_2\\
&= \frac{M+m}{2}\dot{r}^2 + \frac{m}{2} r^2 \dot{\theta}^2 - gr (M - m\cos\theta)\\
\end{align*}

Ahora montamos Euler-Lagrange,

$\displaystyle \frac{d}{dt} \left( \frac{\partial L}{\partial \dot{q}}\right) - \frac{\partial L}{\partial q} = 0$

Para $r$:

$\displaystyle \frac{\partial L}{\partial \dot{r}} = (M+m)\dot{r}$

$\displaystyle \frac{d}{dt} \left( \frac{\partial L}{\partial \dot{r}}\right) = (M+m) \ddot{r}$

$\displaystyle \frac{\partial L}{\partial r} = mr \dot{\theta}^2 -g(M - m\cos \theta)$

Entonces,

$$\boxed{(M+m) \ddot{r} = mr \dot{\theta}^2 -g(M - m\cos \theta)} \tag{1}$$

Para $\theta$:

$\displaystyle \frac{\partial L}{\partial \dot{\theta}} = m r^2 \dot{\theta} $

$\displaystyle \frac{d}{dt} \left( \frac{\partial L}{\partial \dot{\theta}}\right) = m (2r \dot{r}\dot{\theta} + r^2 \ddot{\theta})$

$\displaystyle \frac{\partial L}{\partial \theta} = -grm\sin(\theta)$

Entonces,

$$m \left(2r \dot{r}\dot{\theta} + r^2 \ddot{\theta}\right) = -grm\sin\theta$$

$$2r \dot{r}\dot{\theta} + r^2 \ddot{\theta}= -gr\sin\theta$$

$$\boxed{2 \dot{r}\dot{\theta} + r \ddot{\theta}= -g\sin\theta} \tag{2}$$

**Parte b**

Consideremos que $\sin \theta \approx \theta$ y como $m_{\text{izq}}$ parte del reposo, $\dot{r}(0) \approx 0$. Entonces (2) queda, 

$$ \ddot{\theta} + \frac{g}{r} \theta = 0 \tag{2.b}$$

Cuya sol. es

$$ \theta(t) = \epsilon \cos(\omega t), \quad \omega^2 = g/r$$

Como las masas son iguales, (1) queda,

$$2m \ddot{r} = mr \dot{\theta}^2 -gm(1 - \cos \theta)$$

$$2 \ddot{r} = r \dot{\theta}^2 -g(1 - \cos \theta) \tag{1.b}$$

Entonces,

\begin{align*}
2 \langle \ddot{r} \rangle &= r \langle \dot{\theta}^2 \rangle -g(\langle 1 - \cos \theta \rangle)\\
&= r \langle (-\epsilon \omega \sin(\omega t))^2 \rangle -g\left( \left\langle 1 - \left(1 - \frac{\theta^2}{2}\right) \right\rangle \right)\\
&= r \epsilon ^2 \omega ^2 \langle \sin^2 (\omega t) \rangle -g \frac{\langle \theta^2 \rangle}{2}, \quad \text{(los promedios de $\sin^2$ y $\cos^2$ son $1/2$})\\
&= r \epsilon ^2 \frac{g}{r}\frac{1}{2} - g \frac{\epsilon ^2 \langle \cos^2 (\omega t) \rangle}{2}\\
&= \frac{\epsilon^2 g}{2} - g \frac{\epsilon ^2}{4}\\
\langle \ddot{r} \rangle &= \frac{\epsilon^2 g}{4} -  \frac{g \epsilon ^2}{8}\\
\langle \ddot{r} \rangle &= \frac{g \epsilon ^2}{8}\\
\end{align*}

En nuestro sistema, $r$ es la longitud de lado derecho de la cuerda, osea, la distancia desde la polea hasta el bicho que está oscilando. Como $\langle \ddot{r} \rangle > 0$,  significa que $r$ va aumentando, entonces, el bicho de la izquierda va para arriba.

# **Problema 4** Oscilador de van der Pol

$$ \ddot{x} + \mu (x^2 - a^2) \dot{x} + \omega_0^2 x = 0 \tag{4.19}$$

Con $a = \omega = 1$ nos queda, 

$$ \ddot{x} + \mu (x^2 - 1) \dot{x} + x = 0 \tag{4.20}$$

Reordenando,

$$ \ddot{x} = - \mu (x^2 - 1) \dot{x} - x \tag{4.20.b}$$

De: https://github.com/mbrenesn/FisicaComputacional/blob/main/Tema_13/ODE.ipynb

> ## Ecuaciones Diferenciales de Mayor Orden
>
>La idea es reducir el orden de la ecuación diferencial a cambio de resolver **ecuaciones diferenciales simultáneas**. Por >ejemplo, la ecuación
>$$
\frac{\text d^2 x}{\text d t^2} = f(x, \frac{\text d x}{\text d t}, t),
$$
es equivalente a resolver de forma s
imultánea
>* $\frac{\text d x}{\text d t} = y$,
>* $\frac{\text d y}{\text d t} = f(x, y, t)$.
>
>Estas ecuaciones simultáneas se pueden resolver con el método deseado (Euler, RK2, RK4) para un sistema multivariable. 

Pasemos  $\ddot{x} = - \mu (x^2 - 1) \dot{x} - x$ a un sist. de ec. diff. acoplado para luego meterle Runge-Kutta 4.

$ \dot{x} = v$

$ \dot{v} = \ddot{x} = f(x, v, t)   $

Entonces (4.20.b) queda,

$$ \dot{v} = - \mu (x^2 - 1) v - x \tag{4.20.b'}$$


## Runge-Kutta 4

Definimos el vector de estado:

$$
\mathbf{y}
=\begin{pmatrix}
x \\
v
\end{pmatrix}
$$

y la función:

$$
\mathbf{f}(t,\mathbf{y})=
\begin{pmatrix}
\dot{x} \\
\dot{v}
\end{pmatrix}=
\begin{pmatrix}
v \\
-\mu (x^2-1)v-x
\end{pmatrix}
$$

El método de Runge--Kutta de cuarto orden (RK4) avanza según:

$$
\mathbf{k}_1
=
\mathbf{f}(t_n,\mathbf{y}_n)
$$

$$
\mathbf{k}_2
=
\mathbf{f}
\left(
t_n+\frac{h}{2},
\mathbf{y}_n+\frac{h}{2}\mathbf{k}_1
\right)
$$

$$
\mathbf{k}_3
=
\mathbf{f}
\left(
t_n+\frac{h}{2},
\mathbf{y}_n+\frac{h}{2}\mathbf{k}_2
\right)
$$

$$
\mathbf{k}_4
=
\mathbf{f}
\left(
t_n+h,
\mathbf{y}_n+h\mathbf{k}_3
\right)
$$

Luego, la solución se actualiza mediante:

$$
\boxed{
\mathbf{y}_{n+1}
=
\mathbf{y}_n
+
\frac{h}{6}
\left(
\mathbf{k}_1
+
2\mathbf{k}_2
+
2\mathbf{k}_3
+
\mathbf{k}_4
\right)
}
$$

donde $h$ es el paso temporal.

En este caso, RK4 se aplica al vector de estado:

$$
(x,v)
$$

y no únicamente a $x$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# esta funcion computa la ec.(4.20.b')
def f(t, vec_y, mu):
    x = vec_y[0]
    v = vec_y[1]
    x_dot = v
    v_dot = -mu*(x**2 - 1)*v - x # ec.(4.20.b')
    return np.array([x_dot, v_dot])  # vector de las derivadas de las variables dinamicas. Los va a necesitar RK4.
```

```python
#Implementamos Runge-Kutta de orden 4

def rk4(mu, x0, v0, t0 = 0, tf = 100, h = 0.01):
    # mu: parametro del oscilador
    # x0: condicion inicial x(t=0)
    # v0: condicion inicial v(t=0)
    # t0: tiempo inicial ( = 0 por defecto)
    # tf: tiempo final (  = 100 por defecto)
    # h: paso ( = 0.01 por defecto)
    
    t = np.arange(t0, tf + h, h) # arreglo de tiempos: [t0,tf], en pasos de h
    y = np.zeros((len(t), 2)) # inicializa ceros con la matriz y. Luego RK4 la va rellenando.
    # Esta matriz va a almacenar los pares ordenados (x_n, v_n) para cada tiempo t_n.
    
    #   t_n     | x_n | v_n
    # =====================
    #   t0      | x0  | v0
    #   t0 + h  | x1  | v1
    #   t0 + 2h | x2  | v2
    #   ...
    
    y[0] = [x0, v0] # condiciones iniciales (primera fila de la matriz y)

    for n in range(len(t) - 1): #RK4 en accion!
        k1 = f(t[n], y[n], mu)
        k2 = f(t[n] + h/2, y[n] + h*k1/2, mu)
        k3 = f(t[n] + h/2, y[n] + h*k2/2, mu)
        k4 = f(t[n] + h, y[n] + h*k3, mu)

        y[n+1] = y[n] + (h/6)*(k1 + 2*k2 + 2*k3 + k4)
        # actualiza la fila n+1 de la matriz y, segun el metodo de RK4.
        # es decir, calcula el siguiente estado del sistema.

    return t, y[:, 0], y[:, 1]  # vector de t, vector de x, vector de v, todos de dimension nx1
```

```python

def plot_x_t(t, x, mu, x0, v0):

    plt.figure()

    plt.plot(t, x)

    plt.xlabel("t")
    plt.ylabel("x(t)")

    plt.title(
        f"x(t) | mu={mu}, x0={x0}, v0={v0}"
    )

    plt.grid()

    plt.show()
```

```python

def plot_v_t(t, v, mu, x0, v0):

    plt.figure()

    plt.plot(t, v)

    plt.xlabel("t")
    plt.ylabel("v(t)")

    plt.title(
        f"v(t) | mu={mu}, x0={x0}, v0={v0}"
    )

    plt.grid()

    plt.show()
```

```python
def plot_phase(x, v, mu, x0, v0):

    plt.figure()

    plt.plot(x, v)

    plt.xlabel("x")
    plt.ylabel("v")

    plt.title(
        f"Espacio de fase | mu={mu}, x0={x0}, v0={v0}"
    )

    plt.grid()

    plt.show()
```

- $\mu = 0.07,\, x(0) = 1,\, \dot{x}(0) = 0$

```python
# Parámetros
mu = 0.07
x0 = 1
v0 = 0
```

```python
t, x, v = rk4(mu, x0, v0) # t, x, y son los vectores que retorna RK4

plot_x_t(t, x, mu, x0, v0)

plot_v_t(t, v, mu, x0, v0)

plot_phase(x, v, mu, x0, v0)
```
